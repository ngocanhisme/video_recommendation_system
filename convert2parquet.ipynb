```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("KuaiRand-1K")
    .master("local[*]")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("ERROR")


# =========================
# Input paths (read-only)
# =========================

BASE_PATH = (
    "/kaggle/input/datasets/nguyenngocanhle/"
    "kuairand-1k/KuaiRand-1K/KuaiRand-1K/data"
)

EXTRA_PATH = (
    "/kaggle/input/datasets/nguyenngocanhle/"
    "kuairand-1k"
)


# =========================
# Read CSV files
# =========================

df_rand = spark.read.csv(
    f"{BASE_PATH}/log_random_4_22_to_5_08_1k.csv",
    header=True,
    inferSchema=True
)

df1 = spark.read.csv(
    f"{BASE_PATH}/log_standard_4_08_to_4_21_1k.csv",
    header=True,
    inferSchema=True
)

df2 = spark.read.csv(
    f"{BASE_PATH}/log_standard_4_22_to_5_08_1k.csv",
    header=True,
    inferSchema=True
)

user_features = spark.read.csv(
    f"{BASE_PATH}/user_features_1k.csv",
    header=True,
    inferSchema=True
)

video_features_basic = spark.read.csv(
    f"{BASE_PATH}/video_features_basic_1k.csv",
    header=True,
    inferSchema=True
)

video_features_statistics = spark.read.csv(
    f"{BASE_PATH}/video_features_statistic_1k.csv",
    header=True,
    inferSchema=True
)

video_caption = spark.read.csv(
    f"{EXTRA_PATH}/kuairand_video_captions.csv",
    header=True,
    inferSchema=True
)

video_categories = spark.read.csv(
    f"{EXTRA_PATH}/kuairand_video_categories.csv",
    header=True,
    inferSchema=True
)


# =========================
# Output paths (writable)
# =========================

OUTPUT_PATH = "/kaggle/working/kuairand_parquet"

spark.sql("SET spark.sql.parquet.compression.codec=snappy")


# =========================
# Convert CSV → Parquet
# =========================

df_rand.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/log_random_4_22_to_5_08_1k"
)

df1.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/log_standard_4_08_to_4_21_1k"
)

df2.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/log_standard_4_22_to_5_08_1k"
)

user_features.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/user_features_1k"
)

video_features_basic.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/video_features_basic_1k"
)

video_features_statistics.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/video_features_statistic_1k"
)

video_caption.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/kuairand_video_captions"
)

video_categories.write.mode("overwrite").parquet(
    f"{OUTPUT_PATH}/kuairand_video_categories"
)
```

```python
import os
import subprocess
from IPython.display import FileLink, display

def download_file(path, download_file_name):
    os.chdir('/kaggle/working/')
    zip_name = f"/kaggle/working/{download_file_name}.zip"
    command = f"zip {zip_name} {path} -r"
    result = subprocess.run(command, shell=True, capture_output=True, text=True)
    if result.returncode != 0:
        print("Unable to run zip command!")
        print(result.stderr)
        return
    display(FileLink(f'{download_file_name}.zip'))

download_file("/kaggle/working/kuairand_parquet", "KuiRand-1K-parquet")
```
